## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
import random

def simular_lanzamiento_dados(num_simulaciones=1000):
    """
    Simula el lanzamiento de dos dados 'num_simulaciones' veces y estima
    la probabilidad de que la suma sea mayor que 7 O un número par.

    Args:
        num_simulaciones (int): El número de veces que se realiza el experimento.

    Returns:
        float: La probabilidad estimada.
        str: El resultado de la simulación.
    """
    # Inicializar el contador para los eventos que cumplen la condición
    conteo_eventos_favorables = 0

    # Iterar el experimento el número de veces especificado
    for _ in range(num_simulaciones):
        # 1. Lanzar el primer dado (valor entre 1 y 6)
        dado1 = random.randint(1, 6)
        
        # 2. Lanzar el segundo dado (valor entre 1 y 6)
        dado2 = random.randint(1, 6)
        
        # 3. Calcular la suma total
        suma = dado1 + dado2
        
        # 4. Verificar si la suma cumple la condición:
        #    "mayor a 7 O un número par"
        
        # Condición A: La suma es mayor que 7
        es_mayor_a_7 = suma > 7
        
        # Condición B: La suma es un número par (el residuo de la división por 2 es 0)
        es_par = suma % 2 == 0
        
        # La condición total es A OR B
        if es_mayor_a_7 or es_par:
            conteo_eventos_favorables += 1

    # 5. Calcular la probabilidad estimada
    # (Eventos Favorables / Número Total de Simulaciones)
    probabilidad_estimada = conteo_eventos_favorables / num_simulaciones
    
    # Formatear el resultado para que sea legible
    resultado = (
        f"**Resultados de la Simulación ({num_simulaciones} lanzamientos):**\n"
        f"* Eventos favorables (Suma > 7 O Suma par): **{conteo_eventos_favorables}**\n"
        f"* Probabilidad Estimada: **{conteo_eventos_favorables} / {num_simulaciones} = {probabilidad_estimada:.4f}**"
    )

    return probabilidad_estimada, resultado

NUM_ITERACIONES = 1000
_, resultado_simulacion = simular_lanzamiento_dados(NUM_ITERACIONES)

print(resultado_simulacion)
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Toas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
import random

NUM_BOLAS_BLANCAS = 10
NUM_BOLAS_ROJAS = 20
NUM_BOLAS_VERDES = 30
NUM_TOTAL_BOLAS = NUM_BOLAS_BLANCAS + NUM_BOLAS_ROJAS + NUM_BOLAS_VERDES
NUM_EXTRACCIONES = 5
NUM_SIMULACIONES = 1000


# Crear una lista que representa la caja de bolas (más fácil de muestrear)
caja_de_bolas = (
    ["Blanca"] * NUM_BOLAS_BLANCAS +
    ["Roja"] * NUM_BOLAS_ROJAS +
    ["Verde"] * NUM_BOLAS_VERDES
)

conteo_caso_1 = 0  # Para el evento: 3 Blancas y 2 Rojas
conteo_caso_2 = 0  # Para el evento: Todas del mismo color

# --- Función de Simulación ---
def simular_extracciones(num_simulaciones):
    global conteo_caso_1, conteo_caso_2
    
    for _ in range(num_simulaciones):

        extracciones = random.choices(caja_de_bolas, k=NUM_EXTRACCIONES)
        
        conteo = {
            "Blanca": extracciones.count("Blanca"),
            "Roja": extracciones.count("Roja"),
            "Verde": extracciones.count("Verde"),
        }
        

        if (conteo["Blanca"] == 3 and conteo["Roja"] == 2 and conteo["Verde"] == 0):
            conteo_caso_1 += 1
            
  
        if (conteo["Blanca"] == 5 or conteo["Roja"] == 5 or conteo["Verde"] == 5):
            conteo_caso_2 += 1

simular_extracciones(NUM_SIMULACIONES)

prob_estimada_caso_1 = conteo_caso_1 / NUM_SIMULACIONES
prob_estimada_caso_2 = conteo_caso_2 / NUM_SIMULACIONES


print(f"## 📊 Resultados de la Simulación ({NUM_SIMULACIONES} Iteraciones)")
print("---")

print("### 1. Probabilidad de tomar 3 Blancas y 2 Rojas")
print(
    f"* **Eventos Favorables:** {conteo_caso_1} veces\n"
    f"* **Probabilidad Estimada:** {conteo_caso_1} / {NUM_SIMULACIONES} = **{prob_estimada_caso_1:.5f}**"
)

print("\n### 2. Probabilidad de tomar todas del mismo color")
print(
    f"* **Eventos Favorables:** {conteo_caso_2} veces\n"
    f"* **Probabilidad Estimada:** {conteo_caso_2} / {NUM_SIMULACIONES} = **{prob_estimada_caso_2:.5f}**"
)
```
